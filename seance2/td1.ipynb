```python
import math
# Données
x1 = 2
x2 = 3

w1 = 0.5
w2 = -0.2

b = 0.1
```

```python
# 1. Calcul de z
z = w1 * x1 + ( w2 * x2 + b)
```

```python
# 2. Fonction sigmoïde
def sigmoid(z):
    return 1 / (1 + math.exp(-z))

y = sigmoid(z)
```

```python
print("z =", z)
print("Sortie du neurone =", y)
```

### Rôle du biais

Le biais \(b\) permet de déplacer le seuil d'activation du neurone. Sans biais, la séparation doit passer par l'origine.


###  Rôle des poids

Les poids \(w_1,w_2\) déterminent l'importance de chaque feature dans le calcul :

$$ z=w_1x_1+w_2x_2+b $$

Ici, \(w_1=0.5\) donne une influence positive à \(x_1\), tandis que \(w_2=-0.2\) donne une influence négative à \(x_2\).
