```python
import numpy as np

# Données d'entrée
X = np.array([
    [0, 0],
    [1, 0],
    [0, 1],
    [1, 1]
])

# Sorties attendues
y = np.array([0, 1, 1, 1])

# Poids et biais
w = np.array([0.0, 0.0])
b = 0.0

# Fonction sigmoïde
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Apprentissage
learning_rate = 0.2

for epoch in range(10000):

    # Propagation avant
    z = X @ w + b
    y_pred = sigmoid(z)

    # Erreur
    error = y_pred - y

    # Mise à jour des poids
    dw = X.T @ error / len(X)
    db = np.mean(error)

    w -= learning_rate * dw
    b -= learning_rate * db

print("Poids :", w)
print("Biais :", b)

# Test
predictions = sigmoid(X @ w + b)

print("\nRésultats :")
for inputs, prediction in zip(X, predictions):
    print(inputs, "->", prediction)
```

```python
import math

# Données : OR
X = [
    [0, 0],
    [1, 0],
    [0, 1],
    [1, 1]
]

y = [0, 1, 1, 1]

# Initialisation
w1 = 0.0
w2 = 0.0
b = 0.0

lr = 0.2


def sigmoid(x):
    return 1 / (1 + math.exp(-x))


# Training
for epoch in range(10000):

    for i in range(len(X)):

        x1 = X[i][0]
        x2 = X[i][1]
        target = y[i]

        # Forward
        z = w1*x1 + w2*x2 + b
        prediction = sigmoid(z)

        # Erreur
        error = prediction - target

        # Backpropagation
        dz = error * prediction * (1 - prediction)

        dw1 = dz * x1
        dw2 = dz * x2
        db = dz

        # Update
        w1 -= lr * dw1
        w2 -= lr * dw2
        b -= lr * db


# Test
print("w1 =", w1)
print("w2 =", w2)
print("b  =", b)

print("\nPredictions:")

for x in X:
    z = w1*x[0] + w2*x[1] + b
    prediction = sigmoid(z)

    print(x, "=>", round(prediction, 3))
```

```python
import math

# =========================
# 1. Données
# =========================

X = [
    [0, 0],
    [1, 0],
    [0, 1],
    [1, 1]
]

y = [0, 1, 1, 1]


# =========================
# 2. Modèle
# =========================

w1 = 0.0
w2 = 0.0
b = 0.0

learning_rate = 0.2


def sigmoid(x):
    return 1 / (1 + math.exp(-x))


# =========================
# 3. Training
# =========================

def train(X, y, epochs):

    global w1, w2, b

    for epoch in range(epochs):

        for i in range(len(X)):

            x1 = X[i][0]
            x2 = X[i][1]
            target = y[i]

            # Forward
            z = w1 * x1 + w2 * x2 + b
            prediction = sigmoid(z)

            # Erreur
            error = prediction - target

            # Gradient
            dz = error * prediction * (1 - prediction)

            dw1 = dz * x1
            dw2 = dz * x2
            db = dz

            # Update
            w1 -= learning_rate * dw1
            w2 -= learning_rate * dw2
            b -= learning_rate * db


# =========================
# 4. Predict
# =========================

def predict(x1, x2):

    z = w1 * x1 + w2 * x2 + b

    probability = sigmoid(z)

    if probability >= 0.5:
        return 1
    else:
        return 0


# =========================
# 5. Training
# =========================

train(X, y, 10000)


# =========================
# 6. Test
# =========================

print("Weights:")
print("w1 =", w1)
print("w2 =", w2)
print("b  =", b)

print("\nPredictions:")

for x in X:
    result = predict(x[0], x[1])
    print(x, "=>", result)


# =========================
# 7. Nouvelle donnée
# =========================

print("\nNew prediction:")

print("[0.8, 0.2] =>", predict(0.8, 0.2))
print("[0.1, 0.1] =>", predict(0.1, 0.1))
```

```python
# =========================
# Fonction d'activation
# =========================

def step(x):
    if x >= 0:
        return 1
    return 0


# =========================
# Classe Perceptron
# =========================

class Perceptron:

    def __init__(self, learning_rate=0.1, epochs=10):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = []
        self.bias = 0

    # -------------------------
    # Entraînement
    # -------------------------

    def fit(self, X, y):

        # Initialiser les poids à 0
        n_features = len(X[0])
        self.weights = [0] * n_features

        for epoch in range(self.epochs):

            for xi, target in zip(X, y):

                # Somme pondérée
                z = self.bias

                for i in range(len(xi)):
                    z += self.weights[i] * xi[i]

                # Activation
                prediction = step(z)

                # Erreur
                error = target - prediction

                # Mise à jour des poids
                for i in range(len(xi)):
                    self.weights[i] += self.learning_rate * error * xi[i]

                # Mise à jour du biais
                self.bias += self.learning_rate * error

    # -------------------------
    # Prédiction
    # -------------------------

    def predict(self, X):

        predictions = []

        for xi in X:

            z = self.bias

            for i in range(len(xi)):
                z += self.weights[i] * xi[i]

            prediction = step(z)

            predictions.append(prediction)

        return predictions
```

```python
# Données
X = [
    [0, 0],
    [1, 0],
    [0, 1],
    [1, 1]
]

y = [0, 1, 1, 1]


# Créer le perceptron
model = Perceptron(
    learning_rate=0.1,
    epochs=10
)


# Training
model.fit(X, y)


# Prediction
predictions = model.predict(X)


# Résultats
print("Weights:", model.weights)
print("Bias:", model.bias)

for x, prediction in zip(X, predictions):
    print(x, "=>", prediction)
```
